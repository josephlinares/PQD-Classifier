# Dataset visualization

Dataset comprehends the 16 PQD types according to IEEE 1159, this is a graphical representation of each one

```python
import numpy as np
from matplotlib import pyplot as plt

import config
import dataset
import models
import training
import utils
```

```python
settings = config.load_config('../configs/base_config.yaml')
X, Y_encoded, _, _, encoder, metadata = dataset.loadMatlabDataset(settings['dataset'])
Y = encoder.inverse_transform(Y_encoded)
```

```python
unique_classes = np.unique(Y)

plt.figure(figsize=(10, 6))

for i, class_name in enumerate(unique_classes[:16]):
    idx = np.where(Y == class_name)[0][0]

    plt.subplot(4, 4, i + 1)
    plt.plot(X[idx])
    plt.title(str(class_name))
    plt.xticks([])
    plt.yticks([])

plt.suptitle("Representative Signal per PQD Class", fontsize=14)
plt.tight_layout()
plt.show()
```
