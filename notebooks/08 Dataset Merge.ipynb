# 14 PQD type dataset merge

Current model struggles to classify interruption and sag events, 14 custom datasets for each sample were created and this notebook merges them

```python
import numpy as np
from scipy.io import loadmat, savemat
```

```python
dataset_filename = '../data/' + str(1) + 'pqd_2000pattern_60hz_10cycle_withNoise.mat'
DataBase = loadmat(dataset_filename)
DataBase['SignalsDataBase'][0].shape[0]
```

```python
num_of_pqd = 14
X = [None]
Y = [None]

for i in range(num_of_pqd):
    dataset_filename = '../data/' + str(i) + 'pqd_2000pattern_60hz_10cycle_withNoise.mat'
    DataBase = loadmat(dataset_filename)
    dataset_length = DataBase['SignalsDataBase'][0].shape[0]
    for i in range(dataset_length):
        X.append(DataBase ['SignalsDataBase'][0][i]['signals'][0])
        Y.append(DataBase ['SignalsDataBase'][0][i]['labels'][0])

X = X[1:]
Y = Y[1:]

print("Total samples:", len(X))
```

```python
num_of_pqd = 14
X = []

for i in range(num_of_pqd):
    dataset_filename = '../data/' + str(i) + 'pqd_2000pattern_60hz_10cycle_withNoise.mat'
    DataBase = loadmat(dataset_filename)
    X.extend(DataBase['SignalsDataBase'][0])
    
dtype = DataBase['SignalsDataBase'].dtype

newDataBase = np.array([X], dtype=dtype)
```

```python
dtype = np.dtype([('signals', 'O'), ('labels', 'O')])
newDataBase = np.empty((1, len(X)), dtype=dtype)

for i in range(len(X)):
    newDataBase[0, i] = {
        'signals': X[i],
        'labels': Y[i]
}
```

```python
savemat("../data/14pqd_2000pattern_60hz_10cycle_withNoise.mat", {"SignalsDataBase": newDataBase}, do_compression=True)
```
